## 7.8 Символ слэш (/) в параметрах функции

#### * (звёздочка) - отделяет именнованные параметры функции от всех остальных;
#### / (слэш) - отделяет позиционные параметры функции от всех остальных.

```python
def triangle(a, b, c):
    return a + b + c

res1 = triangle(10, 20, 30)
res2 = triangle(b = 10, a = 20, c=30)
res3 = triangle(10, c=20, b=30)
```

```python
print(res1)
print(res2)
print(res3)
```

Запретим вызов функций с передачей именованных аргументов a, b, c

```python
def triangle(a, b, c, /):
    return a + b + c

res1 = triangle(10, 20, 30)
print(res1)
```

Такой вызов приведёт к ошибкам:

```python
res2 = triangle(b=10,a=20, c=30)
res3 = triangle(10, c=20, b=30)
```

Однако сами по себе параметры функции могут быть определены \
как именнованные:

```python
def triangle(a, b, c=4, /):
    return a + b + c
```

```python
res1 = triangle(10, 20, 30)
res2 = triangle(10, 20)
print(res1)
print(res2)
```

Но по прежнему здесь нельзя указывать именнованные аргументы

```python
res3 = triangle(10, 20, c=30)
```

А вот после символа ' / ' можно прописывать разные параметры \
и передовать в них как позиционные так и именнованные аргументы.

```python
def triangle(a, b, c, /, mul, offset=0):
    return (a + b + c) * mul + offset

res1 = triangle(10, 20, 30, 2)
res2 = triangle(10, 20, 30, offset=1, mul=1)
res3 = triangle(10, 20, 30, 1, 10)
print(res1)
print(res2)
print(res3)
```

То есть, теперь при вызове функции triangle() ей нужно передовать первые \
три позиционных аргумента, а остальные два могут использоваться произвольным образом: \
и как позиционные, и как именнованные.

```python
def polyline(a, b, /, *args, mul=1, offset=0, **kwargs):
    return (a + b + sum(args)) * mul + offset
```

Произвольные набор позиционных аргументов *args должен указываться \
строго после символа слэш, т.к. накладывать на *args какие-либо ограничения \
не имеет смысла, потомучто в эту коллекцию попадают только позиционные аргументы. \
А коллекция **kwargs должна прописываться в самом конце.

```python
res1 = polyline(0, 5, 4, -3, 8)
res2 = polyline(0, 5, 4, -3, mul=2)
res3 = polyline(0, 5, offset=1, mul=2)
res4 = polyline(0, 5, offset=1, mul=2, verbose=True)
```

```python
print(res1)
print(res2)
print(res3)
print(res4)
```

## Совместное использование символов / и *

```python
def rect(width, height, /, offset, *, perimetr=True, **kwargs):
    if perimeter:
        return (width + height) * 2 + offset
    else:
        return width * height + offset
```

В первые два параметра width, height можно передавать только \
позиционные аргументы, в параметр offset - любые, \
а в параметр perimetr - только именнованные.

```python
res1 = rect(10, 20, 0)
res2 = rect(10, 20, perimetr=False, offset=5)
res3 = rect(10, 20, 10, perimetr=False)
```

```python
print(res1)
print(res2)
print(res3)
```

```python
def is_right_tr(a, b, c, /, precision=0.001):
    if abs(c**2 - (a**2 + b**2)) < precision:
        return True
    if abs(a**2 - (b**2 + c**2)) < precision:
        return True
    if abs(b**2 - (a**2 + c**2)) < precision:
        return True
    return False

side_a, side_b, side_c = map(float, input().split())
result = is_right_tr(side_a, side_b, side_c)
print(result)
```

```python
def verify_password(pwd, /, chars="@#!*", min_length=8):
    letters = "абвгдеёжзийклмнопрстуфхцчшщъыьэюяАБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ"
    if len(pwd) >= min_length and set(pwd) & set(chars) and not(set(pwd) & set(letters)):
        return True
    else:
        return False

password = input()
result = verify_password(password, chars="0123456789", min_length=10)
print(result)
```

```python
DEBUG = 10, 'DEBUG'
INFO = 20, 'INFO'
WARNING = 30, 'WARNING'
ERROR = 40, 'ERROR'
CRITICAL = 50, 'CRITICAL'

def log_event(timestamp, message, /, *, 
              level=INFO, format_log="[%(time)] %(levelname) - %(message)"):
    if level[0] < INFO[0]:
        return None

    result = format_log.replace('%(time)', str(timestamp))
    result = result.replace('%(message)', message)
    result = result.replace('%(levelname)', level[1])
    result = result.replace('%(levelno)', str(level[0]))

    return result

log_time = int(input())
log_msg = input()

log_item = log_event(log_time, log_msg, level = WARNING,
                     format_log="%(levelname) - (%(time)) %(message)")

print(log_item)
```

```python
def is_right_rect(a, b, c, d, /, precision=0.001):
    crds = (d, a, b, c, d, a)
    for i in range(1, len(crds) - 1):
        v1 = (crds[i + 1][0] - crds[i][0], crds[i + 1][1] - crds[i][1])
        v2 = (crds[i - 1][0] - crds[i][0], crds[i - 1][1] - crds[i][1])
        len_v1 = (v1[0] * v1[0] + v1[1] * v1[1]) ** 0.5
        len_v2 = (v2[0] * v2[0] + v2[1] * v2[1]) ** 0.5
        cos_a = (v1[0] * v2[0] + v1[1] * v2[1]) / (len_v1 * len_v2)
        if abs(cos_a) > precision:
            return False

    return True
    
rect_coords = [(float(x.split('=')[0]), float(x.split('=')[1])) for x in input().split()]
result = is_right_rect(*rect_coords)
print(result)
```

Проверка через равенство диагоналей

$$
\underbrace{(a_0-c_0)^2 + (a_1-c_1)^2}_{\text{квадрат длины диагонали } AC} - 
\underbrace{(b_0-d_0)^2 + (b_1-d_1)^2}_{\text{квадрат длины диагонали } BD} = 0
$$

```python
def is_right_rect(a, b, c, d, /, *, precision=0.001):
    return abs((a[0]-c[0]) ** 2 - (b[0]-d[0]) ** 2 + (a[1]-c[1]) ** 2 - (b[1] - d[1]) ** 2) < 100 * precision


rect_coords = [(float(x.split('=')[0]), float(x.split('=')[1])) for x in input().split()]

result = is_right_rect(*rect_coords)
print(result)
```

$$
\begin{aligned}
& \text{Проверка равенства диагоналей прямоугольника:} \\
& \\
& \underbrace{(a_x-c_x)^2 + (a_y-c_y)^2}_{\text{диагональ } AC} - 
  \underbrace{(b_x-d_x)^2 + (b_y-d_y)^2}_{\text{диагональ } BD} \approx 0 \\
& \\
& \text{В коде:} \\
& \text{abs((a[0]-c[0])**2 - (b[0]-d[0])**2 + (a[1]-c[1])**2 - (b[1]-d[1])**2) < 100 * precision}
\end{aligned}
$$

\hline
\text{Что вычисляет} & |AC|^2 - |BD|^2 \\
\hline

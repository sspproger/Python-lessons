## 5.2 Операторы break, continue и else

* break - досрочное завершение цикла
* continue - пропуск одной итерации цикла

```python
print("Запуск цикла")

i = 0
while True:
    i += 1
    break
print("Завершение цикла")
```

```python
d = [1, 5, 3, 6, 0, -4]
flFind = False
i = 0

while i < len(d):
    flFind = d[i] % 2 == 0
    if flFind:
        break

    i += 1
print(flFind)
```

```python
i = 0
while i < len(d) and d[i] % 2 != 0:
    i += 1

flFind = i != len(d)
```

```python
s = 0
d = 1
while d != 0:
    d = int(input("Введите целое число: "))
    if d % 2 == 0:
        continue

    s += d
    print("s = " + str(s))
```

```python
S = 0
#i = -10

i = 10
while i < 100:
    if i == 0:
        break

    S += 1/i
    i += 1
else:
    print("Сумма вычеслена корректно")

print(S)
```

```python
p = [0] * 10
count = 5
while count:
    i = int(input())
    if p[i] == 1:
        continue

    p[i] = 1
    count -= 1
print(*p)
```

```python
composition = 1
while True:
    n = int(input())
    if n == 0:
        break
    elif n < 0:
        continue

    composition *= n
else:
    print(1)
print(composition)
```

```python
lst = input().split()
i = 0
while i < len(lst):
    if len(lst[i]) <= 5:
        print("НЕТ")
        break
    i += 1
else:
    print("ДА")

# print('ДА' if min(list(map(len, input().split()))) > 5 else 'НЕТ')
```

```python
name = input().split()
i = 0
while i < len(name):
    if name[i][0].lower() == name[i][-1].lower():
        print("ДА")
        break
    i += 1
    
else:
    print("НЕТ")
```

```python
n = int(input()) 
if n >= 100:
    print("слишком большое значение n")
else:
    lst = []
    while n:
        if n % 3 == 0 and n % 5 == 0:
            lst.append(n)
        n -= 1
    else:
        print(*lst[::-1])
        
```

```python
n = int(input())
i = 1
while True:
    if i**2 > n:
        print(i)
        break
    i += 1
```

```python
x = int(input())
run = 10
day = 1
while run <= x:
    run *= 1.1
    day += 1
print(day)
```

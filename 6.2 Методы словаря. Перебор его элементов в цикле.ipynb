## 6.2 Методы словаря. Перебор его элементов в цикле

#### Метод dict.fromkeys(список[, значение по умолчанию])

```python
a = dict.fromkeys(["+7", "+6", "+5", "+4"])
print(a)
```

```python
a = dict.fromkeys(["+7", "+6", "+5", "+4"], "country code")
print(a)
```

#### Метод clear(). Очистка словаря удаление всех элементов:

```python
a.clear()
print(a)
```

#### Метод copy(). Создание копии словаря:

```python
d = {True: 1, False: "Ложь", "list": [1, 2, 3], 5: 5}
d2 = d.copy()
d2["list"] = [5, 6, 7]
print(d)
print(d2)
```

Так же копию можно создать с помощью функции dict():

```python
print(d2)
d2 = dict(d)
print(d2)
```

#### Метод get() позволяет получить значение словаря по ключу:

```python
d.get("list")
```

Его отличие от оператора d["list"] в том,
что при указании неверного ключа не возникает ошибка,
а выдается по умолчанию значение None:

```python
print(d.get(3))
```

Это значение можно изменить, указав его вторым аргументом:

```python
print(d.get(3, False))
```

#### Метод dict.setdefault(key[, default])

Возвращает значение, ассоциированное с ключом key и если его нет,
то добавляет в словарь со значением None, либо default - если оно указано:

```python
d = {True: 1, False: "Ложь", "list": [1, 2, 3], 5: 5}
```

```python
d.setdefault("list")
```

```python
print(d)
```

```python
d.setdefault(3)
```

```python
print(d)
```

```python
del d[3]
print(d)
```

```python
d.setdefault(3, "three")
print(d)
```

То есть метод setdefault() создаёт новую запись если ключа нет в словаре

```python
print(d)
```

#### Метод keys() возвращает коллекцию ключей

Метод items() возвращает записи ввиде кортежей: ключ, значение.

Метод update() позволяет обновлять словарь содержимым другого словаря

```python
d = dict(one=1, two=2, three="3", four="4")
d2 = {2: "неудовлетворитель", 3: "удовлетворитель", 'four': "хорошо", 5: "отлично"}
print(d)
print(d2)
```

```python
d.update(d2)
print(d)
```

```python
d = dict(one=1, two=2, three="3", four="4")
d2 = {2: "неудовлетворитель", 3: "удовлетворитель", 'four': "хорошо", 5: "отлично"}
d3 = {**d, **d2}
print(d3)
```

## 7.5 Функции с произвольным числом параметров

```python
def os_path(*args):
    print(args)

os_path("F\\~stepic.org", "Добрый, добрый Python (Питон)", "39\\р39. Функции.docx")
```

#### ' * ' - В параметрах функции это оператор упаковки аргументов в кортеж
#### и через переменную args можно с ним работать

```python
def os_path(*args):
    path = "\\".join(args)
    return path

p = os_path("F:\\~stepic.org", "Добрый, добрый Python (Питон)", "39\\р39. Функции.docx")
print(p)
```

```python
p = os_path("F\\~stepic.org", 
            "Добрый, добрый Python (Питон)",
            "39\\р39. Функции.docx",
            sep='/')
```

Ошибка, что функция не имеет такого параметра. \
Дело в том, что записывая объявление *args мы определяем \
лишь произвольное число позиционных параметров, \
но не параметров в виде пары имя=значение.

```python
def os_path(*args, sep='\\'):
    path = sep.join(args)
    return path

p = os_path("F:\\~stepic.org", 
            "Добрый, добрый Python (Питон)",
            "39\\р39. Функции.docx",
            sep='/', trim=True)
```

#### Как определить в функции произвольное число именнованных параметров.

```python
def os_path(*args, **kwargs):
    print(kwargs)
    path = kwargs['sep'].join(args)
    return path
```

Мы прописываем две звёздочки, а затем имя переменной \
которое будет ссылаться на упакованные значения ввиде словаря. 

```python
p = os_path("F:\\~stepic.org", 
            "Добрый, добрый Python (Питон)",
            "39\\р39. Функции.docx",
            sep='/', trim=True)
```

Некоторы параметры можно указать явно:

```python
def os_path(*args, sep='\\', **kwargs):
    path = sep.join(args)
    return path
```

тем самым гарантируем их существование внутри функции, \
а другие передаваемые именнованные аргументы, следует проверять, \
прежде чем использовать.

```python
def os_path(*args, sep='\\', **kwargs):
    if 'trim' in kwargs and kwargs['trim']:
        args = [x.strip() for x in args]

    path = sep.join(args)
    return path
```

Если условие выполняется, то удаляем пробелы до и после \
фрагментов путей к файлу.

```python
p = os_path("  F:\\~stepic.org ", 
            " Добрый, добрый Python (Питон)   ",
            " 39\\р39. Функции.docx ",
            sep='/', trim=True)
print(p)
```

То же самое и с позиционными параметрами. \
Некоторые из них можно указать явно при объявлении функции.

```python
def os_path(disk, *args, sep='\\', **kwargs):
    args = (disk,) + args

    if 'trim' in kwargs and kwargs['trim']:
        args = [x.strip() for x in args]

    path = sep.join(args)
    return path

p = os_path("F:", "~stepic.org", 
            "Добрый, добрый Python (Питон)",
            "39\\p39. Функции.docx",
            sep='/', trim=True)

print(p)
```

И тогда на первый аргумент будет ссылаться параметр disk, а остальные \
позиционные аргументы упаковываться в коллекцию args.

```python
def get_even(*args):
    return [i for i in args if i % 2 == 0]

numbers = list(map(int, input().split()))
print(*get_even(*numbers))
```

```python
def get_biggest_city(*args):
    biggest = args[0]
    for city in args:
        if len(city) > len(biggest):
            biggest = city
    
    return biggest

    
cities = input().split()
print(get_biggest_city(*cities))
```

```python
def get_biggest_city(*lst):
    return max(lst, key=len)

cities = input().split()
print(get_biggest_city(*cities))
```

параметр "key= " в функции max() позволяет указать функцию \
для преобразования элементов перед сравнением, \
То есть для каждого элемента из lst применяется функция указанная \
в key= в данном случае len

```python
def get_data_fig(*args, **kwargs):
    t = (sum(args),)
    if 'tp' in kwargs: 
        t += (kwargs['tp'],)
    if 'color' in kwargs: 
        t += (kwargs['color'],)
    if 'closed' in kwargs: 
        t += (kwargs['color'],)
    if 'width' in kwargs: 
        t += (kwargs['width'],)

    return t

print(get_data_fig(1, 2, 3, 4, tp=True, color=255))
print(get_data_fig(1, 2, tp=False, closed=True, width=5.5, color=100))
```

```python
def get_data_fig(*args, **kwargs):
    result = (sum(args), )
    for key in ('tp', 'color', 'closed', 'width'):
        if key in kwargs:
            result += (kwargs[key], )

    return result


print(get_data_fig(1, 2, 3, 4, tp=True, color=255))
print(get_data_fig(1, 2, tp=False, closed=True, width=5.5, color=100))
```

```python
def str_min(s1, s2):
    return s1 if s1[:2] < s2[:2] else s2

def str_min3(s1, s2, s3):
    return str_min(str_min(s1, s2), s3)

def str_min4(s1, s2, s3, s4):
    return str_min(str_min(str_min(s1, s2), s3), s4)

print(str_min("молоко", "масло"))
print(str_min3("яблоко","вишня","банан"))
print(str_min4("яблоко","банан","вишня","апельсин"))
```

```python
def str_min(*args):
    """Находим строку с минимальным префиксом из 2 символов
    среди любого количества строк """
    return min(args, key=lambda x: x[:2])

print(str_min("молоко", "масло"))
print(str_min("яблоко","вишня","банан"))
print(str_min("яблоко","банан","вишня","апельсин"))
```

## 6.6 Генераторы множеств и словарей

 [<способ формирования значения> for <счетчик> in <итерируемый объект>]

```python
a = [x ** 2 for x in range(1, 5)]
print(a)
```

```python
a = {x ** 2 for x in range(1, 5)}
print(a)
```

```python
a = {x: x ** 2 for x in range(1, 5)}
print(a)
```

```python
d = [1, 2, '1', '2', -4, 3, 4]
a = {int(x) for x in d}
print(a)
```

Генераторы работают быстрее циклов!

```python
d = [1, 2, '1', '2', -4, 3, 4]
set_d = set()
for x in d:
    set_d.add(int(x))
print(set_d)
```

```python
import timeit

data = list(range(10000))

def set_comprehension():
    return {x**2 for x in data}

def loop_set():
    result = set()
    for x in data:
        result.add(x**2)
    return result

set_comp_time = timeit.timeit(set_comprehension, number=1000)
set_loop_time = timeit.timeit(loop_set, number=1000)
print(f"Генератор множеств: {set_comp_time:.4f} сек")
print(f"Цикл for: {set_loop_time:.4f} сек")
print(f"Генератор быстрее на {(set_loop_time/set_comp_time - 1) * 100:.1f}%" 
if set_comp_time < set_loop_time
else f"Цикл быстрее на {(set_comp_time/set_loop_time - 1)* 100:.1f}%")
```

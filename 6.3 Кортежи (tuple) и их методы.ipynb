## 6.3 Кортежи (tuple) и их методы

Кортеж - это упорядоченная но не изменяемая коллекция произвольных данных.

```python
a = 1, 2
print(a)
```

```python
a = (1, 2, 3)
print(a)
```

```python
b = 1
print(b)
b = 1,
print(b)
b = (1,)
print(b)
```

```python
x, y = (1, 2)
print(x, y)
```

```python
x, y = (1, 2, 3)
```

```python
a = 1, 2, 3
print(a)
```

```python
print(len(a))
```

### Доступ к элементам кортежа осуществляется:
#### по индексу:
* a[0]
* a[2]
* a[-1]
#### через срезы:
* a[1:2]
* a[0:-1]
* a[:3]
* a[1:]
* a[:]

#### Ссылка на один и тот же объект:

```python
b = a[:]
print(id(a), id(b))
```

- Для списков эта операция b = a[:] приводит к их копированию.
- Для кортежей эта операция b = a[:] возвращает тот же самый объект.

```python
a = (1, 2, 3)
a[1] = 100
```

Кортежи можно использовать в качестве ключей у словарей:

```python
d = {}
d[a] = "кортеж"
print(d)
```

Кортеж занимает меньше памяти. \
__sizeof__() возвращает размер данных в байтах.

```python
lst = [1, 2, 3]
t = (1, 2, 3)
print(f"Размер списка в байтах: {lst.__sizeof__()}")
print(f"Размер кортежа в байтах: {t.__sizeof__()}")
```

```python
a = ()
b = tuple()
print(type(a), type(b))
```

#### Мы не можем менять уже существующие элементы в кортеже, 
#### Но можем создавать новые, используя оператор +

```python
a = ()
print(f"a = {a}")
a = a + (1, )
print(f"a = {a}")
```

```python
a = (2, 3) + a
print(f"a = {a}")
```

```python
a += (("a", "hello"))
print(a)
```

```python
b = (0,) * 10
print(b)
b = ("hello", "world") * 5
print(b)
```

Удалять отдельные элементы кортежа нельзя:

```python
del a[1]
```

Целиком можно удалить объект:

```python
a = 1, 2, 3
print(a)
del a
print(a)
```

```python
a = tuple([1, 2, 3])
print(a)
a = tuple("Привет мир")
print(a)
```

```python
t = (1, 2, 3)
lst = list(t)
print(lst)
```

```python
a = (True, [1, 2, 3], "hello", 5, {"house": "дом"})
print(a)
```

```python
print(a[1])
a[1].append("5")
print(a[1])
```

### Методы:
* tuple.count(значение) - возвращает число найденных элементов с указанным значением
* tuple.index(значение, start, stop(значение stop не включается в поиск))
возвращает индекс первого найденного элемента с указанием значений начала и конца поиска.

```python
a = ("abc", 2, [1, 2], True, 2, 5)
a.count("abc")
```

```python
a.count(2)
```

```python
a.count("a")
```

```python
a.index(2)
```

```python
a.index(3)
```

```python
3 in a
```

```python
2 in a
```

```python
a.index(2, 3) # второй параметр - индекс начала поиска
```

```python
a.index(2, 3, 5) # третий параметр - индекс конца поиска
```

```python
a.index(2, 3, 4) # ошибка, в третьем индексе нет значения 2
```

```python
nums = tuple(map(int, input().split()))
t = () 
for n in nums:
    if n not in t:
        t += (n,)
print(*t)
```

```python
nums = tuple(map(int, input().split()))
for i, n in enumerate(nums):
    if n not in nums[i+1:]:
        print(i, end=' ')
        
```

```python
nums = tuple(map(int, input().split()))
counts = {}
for n in nums:
    counts[n] = counts.get(n, 0) + 1
indices = [i for i, n in enumerate(nums) if counts[n] > 1]
print(*indices)
```

```python
nums = tuple(map(int, input().split()))
indices = ()
for i in range(len(nums)):
    if nums.count(nums[i]) > 1:
        indices += (i,)
print(*indices)
```

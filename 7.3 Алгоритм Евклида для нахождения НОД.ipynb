7.3 Алгоритм Евклида для нахождения НОД
Нахождение наибольшего общего делителя (НОД) для двух натуральных чисел a и b.

```python
def get_nod(a, b):
    """Вычисляется НОД для натуральных чисел a и b 
    по алгоритму Евклида.
    Возвращает вычисленный НОД.
    """
    while a != b:
        if a > b:
            a -= b
        else: 
            b -= a
    return a
```

```python
import time

def test_nod(func):
    # -- тест №1 ----------------------
    a = 28
    b = 35
    res = func(a, b)
    if res == 7:
        print("#test1 - ok")
    else:
        print("#test1 - fail")

    # -- тест №2 -----------------------
    a = 100
    b = 1
    res = func(a, b)
    if res == 1:
        print("#test2 - ok")
    else:
        print("#test2 - fail")

    # -- тест №3 ------------------------
    a = 2
    b = 100000000

    st = time.time()
    res = func(a, b)
    et = time.time()
    dt = et - st
    if res == 2 and dt < 1:
        print("#test3 - ok")
    else:
        print("#test3 - fail")

test_nod(get_nod)
```

```python
def get_fast_nod(a, b):
    """Вычисляется НОД для натуральных чисел a и b
       по быстрому алгоритму Евклида.
       Возвращает вычисленный НОД.
    """
    if a < b:
        a, b = b, a

    while b != 0:
        a, b = b, a % b

    return a

test_nod(get_fast_nod)
```

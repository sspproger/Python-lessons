def <имя функции>([список аргументов]): \
        оператор 1 \
        оператор 2 \
        … \
        оператор N

```python
def send_mail():
    text = "Привет!"
    print(text)

send_mail()
```

```python
def send_mail(from_name):
    text = f"""
    Hello, 
    Hello, 
    Hello 
    {from_name}!"""
    print(text)

send_mail('Anton')
```

```python
def send_mail(from_name, old):
    text = f"""Hello {from_name},
мне {old} лет."""
    print(text)

send_mail('Anton', 27)
```

```python
def f():
    print("It's my first function")
f()
```

```python
def f():
    name, surname = input().split()
    print(f"Уважаемый, {name} {surname}! Вы верно выполнили это задание!")

f()
```

```python
def weight(x):
    print(f"Предмет имеет вес: {x} кг.")

x = float(input())
weight(x)
```

```python
def lst(x):
    print(f"Min = {min(x)}, max = {max(x)}, sum = {sum(x)}")

x = list(map(int, input().split()))
lst(x)
```

```python
def perimetr(width, height):
    print(f"Периметр прямоугольника, равен {(width + height) * 2}")

w, h = map(int, input().split())
perimetr(w, h)
```

```python
def check(email):
    allowed = set('abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789_@.')
    if '@' in email and '.' in email and set(email) <= allowed:
        print('ДА')
    else:
        print('НЕТ')

email = input()
check(email)
        
```

```python
import string
def check(email):
    allowed = set(string.ascii_letters + string.digits + '_@.')
    if '@' in email and '.' in email and set(email) <= allowed:
        print('ДА')
    else:
        print('НЕТ')

email = input()
check(email)
```

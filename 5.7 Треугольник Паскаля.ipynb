## 5.7 Треугольник Паскаля

"По краям каждой строки стоят еденицы, а каждое из остальных чисел равно сумме двух стоящих над ним из предыдущей строки." \
Блез Паскаль

* row = [1] * (i + 1)
* j == 0 or j == i
* row[j] = P[i-1][j-1] + P[i-1][j]

```python
N = 12
P = []

for i in range(0, N):
    row = [1] * (i + 1)
    for j in range(i + 1):
        if j != 0 and j != i:
            row[j] = P[i-1][j-1] + P[i-1][j]

    P.append(row)

for r in P:
    print(r)

print()

# находим максимальное число в треугольнике
max_num = max(max(row) for row in P)

# определяем ширину для каждого числа
cell_width = len(str(max_num)) + 1 # +1 для пробела между числами 

for i, r in enumerate(P):
    # количество пробелов перед первой строкой
    s = ' ' * (N - i - 1) * (cell_width // 2 + 1) 
    formatted_row = '  '.join(f'{num:^{cell_width}}' for num in r)
    print(s + formatted_row)
```

Вначале определяем общее количество строк, затем коллекцию P, \
которая будет хранить значения треугольника Паскаля и, далее, \
вложенные циклы. Первый цикл со счётчиком i перебирает строки, \
а второй - со счётчиком j перебирает элементы текущей строки row. \
Внутри вложенного цикла проверяем, если индекс j ссылается не на \
крайние элементы строки, то формируем их значение на основе суммы двух чисел \
из предыдущей строки. После этого добавляем строку в список P и проделываем эту \
операцию N раз. В конце программы выводим результат.


* '^' - выравнивание по центру
* '<' - выравнивание по левому краю
* '>' выравнивание по правому краю

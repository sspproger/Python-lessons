## 6.4 Множества (set) и их методы

#### Множество set() - это неупорядоченная коллекция уникальных элементов. 
#### Уникальных занчит в множестве отсутствуют дублирующие значения.

```python
a = {1, 2, 3, "hello"}
print(a)
print(type(a))
```

#### при попытке в множестве разместить несколько одинаковых значений приведёт к тому -
#### что в множестве останутся только не повторяющиеся значения

```python
a = {1, 2, 3, "hello", 2, 3, "hello"}
print(a)
```

#### Во множеству мы можем записывать только неизменяемые данные:
* числа
* булевые значения
* строки
* кортежи

```python
a = {1, 4.5, "hi", "hi", (4, True), ("a", False)}
print(a)
```

#### Изменяемые типы данных добавлять нельзя:
* списки
* словари
* множества

```python
b = {[1, 2], 3, 4} # list
b = {{1: 1, 2: 2}, 3, 4} # dictionary
b = {{1, 2}, 3, 4} # set
```

Для создания множества используется встроенная функция set()

```python
set()
```

Если попытаться создвать пустое множество так: b = {}, \
то получим пустой словарь.

```python
b = {}
print(type(b))
```

Пустое множество создаётся именно с помощью функции set()

```python
b = set()
print(type(b))
```

Если указать какой-либо итерируемый объект, например, список, \
то получим множество, состоящее из уникальных значений этого списка.

```python
b = set([1, 2, 1, 0, -1, 2])
print(b)
```

Нет ошибки потомучто функция set() принимает только один аргумент \
и создаёт множество из элементов этого списка.

```python
b = set([1, 2, 1, 0, -1, 2], 4, 5)
print(b)
```

```python
b = set("abracadabra")
print(b)
```

```python
b = set(range(7))
print(b)
```

Множества представляют собой не упорядоченную коллекцию, то есть, \
значение элементов могут располагаться в них в любом порядке. \
Это так же означает, что с множествами нельзя выполнять операции доступа \
по индексу или срезам:

```python
print(b[0])
```

## Методы добавления/удаления элементов множества

add() - позволяет добавлять новое значение в множество:

```python
b = set()
print(b)
b.add(7)
print(b)
```

```python
b.add(7) # если такое значение уже есть то оно не будет добавлено
print(b)
```

update() - добавление сразу нескольких значений в множество

```python
print(b)
b.update(["a", "b", (1, 2)])
print(b)
```

```python
b.update("abracadabra")
print(b)
```

duscard() - удаление элементов по значению, если такого элемента нет, \
то ошибка не происходит

```python
print(b)
b.discard(7)
print(b)
```

```python
print(b)
b.discard(7)
print(b)
```

remove() - метод для удаления элемента по значению, если такого элемента нет, \
то возникает ошибка.

```python
print(b)
b.remove('a')
print(b)
```

remove() возвращает ошибку при попытке удаления несуществующего значения.

```python
b.remove('a')
```

pop() - удаляет элемент рандомно по значению и возвращает это значение,  \
если удалить из пустого множества, то будет ошибка.

```python
print(b)
b.pop()
print(b)
```

```python
c = set()
c.pop()
```

clear() - удаляет все элементы из множества

```python
print(b)
b.clear()
print(b)
```

```python
# lst_in = list(map(str.strip, sys.stdin.readlines()))
lst_in = ["Мария", "Елена", "Екатерина", "Александр", "Елена", "Мария"]
print(len(set(lst_in)))
```

```python
lst_in = ['EvgeniyK: спасибо большое!',
          'LinaTroshka: лайк и подписка!',
          'Sergey Karandeev: крутое видео!',
          'Евгений Соснин: обожаю',
          'EvgeniyK: это повтор?',
          'Sergey Karandeev: нет, это новое видео']
d = {}
for i in lst_in:
    name, message = i.split(': ', 1)
    if name not in d:
        d[name] = [message]
    else:
        d[name].append(message)
print(len(set(d)))
```

```python
d = set()
for i in lst_in:
    d.add(i.split(':')[0])
print(len(d)) 
```

```python
print(len({name[0] for name in [c.split(': ') for c in lst_in]}))
```

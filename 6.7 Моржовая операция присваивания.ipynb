## 6.7 Моржовая операция присваивания

#### var := <вычисляемое выражение>
Это именно операция а не оператор, то есть её нельзя прописать, как завершенную, законченную команду языка Python.

```python
d := 77 # ошибка
```

Моржовая операция может быть использована только как составная часть \
какого-либо оператора.

```python
if d := 77:
    print(d)
```

В этом случае будет создана новая переменная d, если ранее она не существовала, \
ей присваивается значение 77, присвоенное значение возвращается можовой операцией, \
и по условию выводится на экран.

Чаще всего справа от моржовой операции указываются не кокретные значения, \
а вычисляемое значение, например, какая-либо функция:

```python
if d := int(input()):
    print(d)
```

Это условие будет отрабатывать, пока пользователь вводит с клавиатуры \
ненулевые целые числа. Если ввести 0, то оно интерпритируется, как False, \
и функция print() не отрабатывает.

```python
if d:= int(input()) > 0:
    print(d)
```

Сначала выполняется чтение целого значение введённого пользователем, \
затем, его сравнение с нулем и только после этого полученное булевое значение \
присваиваестя переменной d.

```python
if (d := int(input())) > 0:
    print(d)
```

Сначала то что в скобочках, поэтому d = 4

## Преимущества моржовой операции присваивания

```python
s = 0
d = int(input())
while d != -1:
    s += d
    d = int(input())
```

```python
s = 0
while (d := int(input())) != -1:
    s += d
```

```python
n = 10
lst = [b for x in range(n) if (b := x * x) < n]
print(*lst)
```

```python
lst2D = [
    row1 := [1, 2, 3],
    row2 := [4, 5, 6],
    row3 := [7, 8, 9],
]
print(*lst2D)
print()
row2[1] = 100
print(*lst2D)
```

## Недостатки моржовой операции присваивания

```python
if (a := int(input())) % 2 == 0 and (b := abs(a)) < 30:
    print(a, b)
else:
    print(b)
```

Переменная b может не существовать при переходе к ветке else. \
Например, если значение a будет не кратно 2, то вторая часть условия по and \
отрабатывать не будет, и, как следствие, переменная b не создасться \
(если её небыло ранее).

```python
var = (d := 77) + 10 / 2 # Таких конструкций следует избегать
```

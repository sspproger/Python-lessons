## 7.4 Позиционные и именованные аргументы. Параметры со значениями

```python
def get_V(a, b, c):
    print(f"a = {a}, b = {b}, c = {c}")
    return a * b * c

v = get_V(1, 2, 3)
print(v)
```

```python
v = get_V( b = 1, a = 2, c = 3)
print(v)
```

Такие аргументы называются "именнованные"

Вначале всегда указываются позиционные элементы, а после именнованные.

```python
v = get_V(1, c=2, b=3)
print(v)
```

Можно задавать параметры со значениями по умолчанию:

```python
def get_V(a, b, c, verbose=True):
    if verbose:
        print(f"a = {a}, b = {b}, c = {c}")

    return a * b * c

v = get_V(1, 2, 3)
print(v)
```

```python
v = get_V(1, 2, 3, verbose = False)
print(v)
```

```python
v = get_V(1, 2, 3, False)
print(v)
```

```python
def compare_str(s1, s2, reg = False, trim = True):
    if reg:
        s1 = s1.lower()
        s2 = s2.lower()
    if trim:
        s1 = s1.strip()
        s2 = s2.strip()

    return s1 == s2
```

```python
print(compare_str("Python ", " Python"))
```

или менять поведения функции указывая другие параметры reg, trim

```python
print(compare_str("Python ", " Python", trim = False))
```

```python
print(compare_str("Python", "PYTHON", True, False))
```

или так:

```python
print(compare_str("Python", "PYTHON", reg = True, trim = False))
```

```python
def add_value(value, lst=[]):
    lst.append(value)
    return lst

l = add_value(1)
l = add_value(2)
print(l)
```

Когда мы объявляем функцию, то создаётся объект-функция \
и объекты значений, которыми инициализируются параметры, \
в данном случае пустой список. \
Затем, когда вызывается функция, то параметр lst ссылается на этот список \
и добавление элемента происходит в него.

Если при вызове функции каждый раз передавать пустой список, \
то параметр lst каждый раз будет ссылаться на новый список \
и каждый раз элемент будет добавляться в пустой список.

```python
l = add_value(1, [])
l = add_value(2, [])
print(l)
```

Чтобы такое поведение было по умолчанию, \
чтобы явно не приходилось передавать пустой список:

```python
def add_value(value, lst=None):
    if lst is None:
        lst = []

    lst.append(value)
    return lst
```

Значение параметра lst инициализируется неизменяемым значением None, \
а в самой функции выполяется проверка: \
если параметр lst равен None, то создаётся новый пустой список \
и в него помещается значение value.

```python
l.clear()
l = add_value(1)
l = add_value(2)
print(l)
```

а если хотим продолжить предыдущий, \
то достаточно его указать вторым аргументом

```python
l = add_value(2, l)
print(l)
```

```python
def check_password(s, chars="$%!?@#"):
    if len(s) < 8:
        return False
    
    for c in s:
        if c in chars:
            return True
    return False

password = "1234534#345!"
print(check_password(password))
```

```python
def html(s, tag='h1'):
    return f'<{tag}>{s}</{tag}>'


s = input()
print(html(s))
print(html(s, tag='div'))
```

```python
def html(s, tag='div', up=True):
    if up:
        return f'<{tag.upper()}>{s}</{tag.upper()}>'
    else:
        return f'<{tag}>{s}</{tag}>'

s = input()
print(html(s))
print(html(s, up=False))
```

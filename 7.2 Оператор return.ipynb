```python
def get_sqrt(x):
    res = None if x < 0 else x ** 0.5

a = get_sqrt(49)
print(a)
```

```python
def get_sqrt(x):
    res = None if x < 0 else x ** 0.5
    return res
    return x # не выполнится так как функция завершает работу после первого return
print(get_sqrt(49))
```

```python
def get_sqrt(x):
    res = None if x < 0 else x ** 0.5
    return res
    print(x)

print(get_sqrt(49))
```

```python
def get_sqrt(x):
    res = None if x < 0 else x ** 0.5
    return (res, x) # на практике часто не пишут круглые скобки, 
    # а просто пишут элементы через запятую.
```

```python
a, b = get_sqrt(49)
print(a, b)
```

```python
def get_max2(a, b):
    return a if a > b else b

x, y = 5, 7
print(get_max2(x, y))
```

```python
x, y, z = 5, 7, 10
print(get_max2(x, get_max2(y, z)))
```

```python
def get_max3(a, b, c):
    return get_max2(a, get_max2(b, c))

print(get_max3(x, y, z))
```

```python
def is_not_even(n):
    return n%2!=0

lst_d = list(map(int, input().split()))
lst = [i for i in lst_d if is_not_even(i)]
print(*lst)
```

```python
def s_len(s):
    return len(s) >= 6

cities = input().split()
lst = [i for i in cities if s_len(i)]
print(*lst)
```

```python
def line(s):
    return s, len(s)

cities = input().split()
d = dict(line(city) for city in cities)
print(sorted(d))
print(*sorted(d, key=d.get))
print(*sorted(d.items()))

#d = {city: length for city, length in [line(city) for city in cities]}
```

## 7.7 Символ звездочка (*) параметрах функции

* *args - для хранения значений позиционных аргументов;
* **kwargs - для хранения пар ключ-значение именованных аргументов;

Функция box_nd. \
Вычисляет периметр многомерного прямоугольного параллелепипеда, \
либо его объём, в зависимости от значения параметра perimeter:

```python
import math
def box_nd(a, b, c, *args, perimeter=True):
    if perimeter:
        return sum((a, b, c, *args)) * 2 ** (len((a,b,c, *args)) - 1)
    else:
        return math.prod((a,b,c,*args))
```

При вызове этой функции обязательно нужно передать первые три аргумента для параметров \
a, b, c и при необходимости, дополнительные, которые помещаются в коллекцию *args.

```python
res1 = box_nd(5, 6, 7)
res2 = box_nd(5, 7, 3, 5, 2)
res3 = box_nd(5, 7, 3, 5, 2, perimeter=False)
print(res1)
print(res2)
print(res3)
```

В последнем вызове значение параметра perimeter \
мы вынуждены передавать через именованный аргумент, \
т.к. если прописать только значение False:

```python
res3 = box_nd(5, 7, 3, 5, 2, False)
```

то оно будет автоматически помещаться в коллекцию *args \
т.к. становится позиционным аргументом. \
В параметры функции, которые записаны после *args, \
можно передавать значения только с помощью именованных аргументов!

```python
def box_nd(a, b, c, *args, perimeter=True, initial, **kwargs):
    if perimeter:
        return sum((a, b, c, *args)) * 2 ** (len((a, b, c, *args)) - 1) + initial
    else:
        return math.prod((a, b, c, *args)) + initial
```

Если не указать в параметрах функции значение именнованного элемента \
(который находится после *args), в данном случае initial, \
то это не приводит к ошибке, так как интерпритатор языка Python воспринимает его \
как именованный элемент у которого просто нет начального значения.

```python
res1 = box_nd(5, 7, 3, initial=2)
res2 = box_nd(5, 7, 3, 5, 2, initial=0, verbose=True)
res3 = box_nd(5, 7, 3, 5, 2, initial=-1, perimeter=False)
print(res1)
print(res2)
print(res3)
```

То есть здесь обязательно нужно передовать именнованный аргумент initial.

Если значение initial будте задано по умолчанию initial=0, \
то при вызове функции его можно будет не прописывать.

```python
def box_nd(a, b, c, *args, perimeter=True, initial=0, **kwargs):
    if perimeter:
        return sum((a, b, c, *args)) * 2 ** (len((a, b, c, *args)) - 1) + initial
    else:
        return math.prod((a, b, c, *args)) + initial
```

Имя *args которое мы постоянно прописывали после звёздочки - это некий стандарт, \
принятый в языке Python. Но никто нам не мешает использовать другое имя.

```python
def box_nd(a, b, c, *ar, perimeter=True, initial=0, **kwargs):
    if perimeter:
        return sum((a, b, c, *ar)) * 2 ** (len((a, b, c, *ar)) - 1) + initial
    else:
        return math.prod((a, b, c, *ar)) + initial
```

Можно вообще убрать символы после звёздочки:

```python
def box_nd(a, b, c, *, perimeter=True, initial=0, **kwargs):
    if perimeter:
        return sum((a, b, c)) * 2 ** (len((a, b, c)) - 1) + initial
    else:
        return math. prod((a, b, c)) + initial
```

Теперь в функции box_nd() нет коллекции, которая сохраняла бы позиционные аргументы. \
Поэтому в теле функции мы не можем с ней никак обращаться. Её пришлось убрать. \
И, кроме того, теперь, при вызове этой функции, можно передавать только первые три аргумента в a, b, c и никакие дополнительные. А вот в остальные параметры функции \
по-прежнему можно передавать только именованные аргументы:

```python
res1 = box_nd(5, c=7, b=3, initial=2) # OK
res2 = box_nd(5, 7, 3) # OK
res2 = box_nd(5, 7, 3, False) # ERROR
res3 = box_nd(5, 7, 3, 5, 2, perimeter=False) # ERROR
```

Таким образом что символ звёздочка ' * ' в параметрах функции \
разделяет параметры на позиционные и именнованные. \

```python
def func(a, b=2, c=3, *, verbose, key=None): ...
```

def func(a, b=2, c=3, *, verbose, key=None):
здесь в параметры функции a, b, c можно передавать произвольные аргументы \
и позиционные и именнованные, а в параметры verbose и key только именнованные.

```python
def func(a, b=2, c=3, *args, verbose, key=None): ...
```

Здесь по-прежнему будет выполняться разделение параметров на обычные \
и строго именованные, но дополнительно в функцию func можно передавать \
произвольное количество позиционных аргументов.

```python
def compare_str(s1, s2, *, ignore_case=True):
    if ignore_case:
        return s1.lower() == s2.lower()
    else:
        return s1 == s2


res1 = compare_str("Python", "python")
res2 = compare_str(s2="Go", s1="go", ignore_case=False)
res3 = compare_str("Java", "java", False)
print(res1)
print(res2)
print(res3)
```

```python
def count_chars(string, chars, *, return_type=tuple, ignore_case=True):
    if ignore_case:
        string = string.lower()
        chars = chars.lower()

    counts = [string.count(char) for char in chars]

    if return_type == tuple:
        return tuple(counts)
    elif return_type == list:
        return list(counts)
    elif return_type == set:
        return set(counts)
    else:
        return tuple(counts)

msg = "Python is an easy to learn, powerful programming language."
print(count_chars(msg, "aouvei", return_type=list, ignore_case=False))
print(count_chars(msg, "aouvei", return_type=set))
print(count_chars(msg, "aouvei"))
```

```python
def symbol_upper(msg, indx=0, *, ignore_indx, to_upper=True):
    if indx in ignore_indx:
        return msg

    new_char = msg[indx].upper() if to_upper else msg[indx].lower()
    msg = msg[:indx] + new_char + msg[indx+1:]

    return msg

res1 = symbol_upper("Python is the best language.", ignore_indx=(3, 6))
res2 = symbol_upper("Python is the best language.", 2, ignore_indx=(3, 6))
res3 = symbol_upper("Python is the best language.", 3, ignore_indx = [6])
print(res1)
print(res2)
print(res3)
```

```python
def filter_by_length(*args, min_length=0, max_length):
    lst = []
    for s in args:
        if min_length <= len(s) <= max_length:
            lst.append(s)
    return lst

result = filter_by_length(
    "Мономах", "Невский", "Радонежский", "Ломоносов", 
    "Лобачевский", min_length=5, max_length=10)
print(result)

# names_result = filter_by_length(*names_initial, min_length=5, max_length=9)
```

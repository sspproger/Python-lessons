## Функция dict()

```python
d2 = dict(one = 1, two = 2, three = "3", four = '4')
print(d2)
```

В функции dict() в ключах числа использовать не получится!

```python
d2 = dict(1 = "one", two = 2, three = "3", four = "4")
print(d2)
```

```python
lst = [[2, "неудовлетворительно"], [3, "удовлетворительно"], [4, "хорошо"], [5, "отлично"]]
print(lst)
```

```python
d3 = dict(lst)
print(d3)
```

```python
d[True] = "Истина"
d[False] = "Ложь"
print(d)
```

```python
d[True] = 1
print(d)
```

* Словарь как и список относится к изменяемым типам данных

Если в качестве ключа указать список т.е. изменяемый тип данных,
то будет ошибка. 
Таким образом в ключах можно указывать только НЕ изменяемые типы данных, 
а в значениях как изменяемые так и не изменяемые.

```python
d[[1,2]] = 1
```

```python
import math
cache = {}
while True:
    num = int(input())
    if num == 0: break
    if num in cache:
        print(f"значение из кэша: {cache[num]}")
    else:
        val = round(math.sqrt(num), 2)
        cache[num] = val
        print(val)
```

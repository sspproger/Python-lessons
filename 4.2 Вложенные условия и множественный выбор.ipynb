#### Оператор elif

```python
item = int(input())

if item == 1:
    print("Выбран курс по Python")
else:
    if item == 2:
        print("Выбран курс по C++")
    else:
        if item == 3:
            print("Выбран курс по Java")
        else:
            if item == 4:
                print("Выбран курс по JavaScript")
            else:
                print("Неверный пункт")
```

```python
m, n = map(int, input().split())
days = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
if n == 1: 
    prev_m = m - 1
    prev_n = days[prev_m]
else:
    prev_m = m
    prev_n = n - 1

if n == days[m]:
    next_m = m + 1
    next_n = 1
else:
    next_m = m
    next_n = n + 1

print(f"{prev_m:02d}.{prev_n:02d} {next_m:02d}.{next_n:02d}")
```

## 5.8 Генераторы списков (List comprehension)

```python
N = 6
a = [0] * N

for i in range(N):
    a[i] = i ** 2

print(a)
```

```python
a = [x**2 for x in range(N)]
print(a)
```

## Условия в генераторах списков

```python
a = [x for x in range(-5, 5) if x <= 0]
print(a)
```

```python
a = [x for x in range(-5, 5) if x % 2 != 0]
print(a)
```

```python
a = [x for x in range(-6, 7) if x % 2 == 0 and x % 3 == 0]
print(a)
```

```python
cities = ["Москва", "Тверь", "Рязань", "Ярославль", "Владимир"]
a = [city for city in cities if len(city) < 7]
print(a)
```

```python
d = [4, 3, -5, 0, 2, 11, 122, -8, 9]
a = ["четное" if x % 2 == 0 else "нечетное" for x in d]
print(a)
```

```python
a = [x + 2 for x in d]
print(a)
```

```python
d = [4, 3, -5, 0, 2, 11, 122, -8, 9]
a = ["четное" if x % 2 == 0 else "нечетное"
    for x in d
    if x > 0
    ]
print(a)
```

```python
lst = [d for d in input().split()]
lst_abs = [abs(float(d)) for d in lst]
print(lst_abs)
```

```python
print([int(d) for d in input()])
```

```python
print(*[d for d in input().split() if len(d) > 5])
```

```python
n = int(input())
print(*[d for d in range(1, n + 1) if n % d == 0])
```

## Является ли число простым или нет

```python
n = int(input())
if n < 2:
    print("НЕТ")
else:
    for i in range(2, n):
        if n % i == 0:
            print("НЕТ")
            break
    else:
            print("ДА")
```

```python
cities = input().split()
res = "ДА"
for i in range(len(cities) - 1):
    last = cities[i][-1]
    if last in 'ьъы':
        last = cities[i][-2]
    first = cities[i + 1][0]

    if last.lower() != first.lower():
        res = "НЕТ"
        break

print(res)
```

```python
n = int(input())
sum = 0
for i in range(n):
    if i % 3 == 0 or i % 5 == 0:
        sum += i
print(sum)
```

<h4 align="center">Основные команды LaTeX:</h4>

|  Команда	  |     Результат	|   Что делает   |
|-------------|-----------------|----------------|
|    x^2	  |     $x^2$	    |     степень    |
|    x_2	  |     $x_2$	    |     индекс     |
|\frac{a}{b}  | $\frac{a}{b}$   |	  дробь      |
|\sqrt{x}	  | $\sqrt{x}$	    |     корень     |
|\sum_{i=1}^n |	$\sum_{i=1}^n$	|     сумма      |
|\int_a^b	  | $\int_a^b$	    |     интеграл   |
|\alpha	      | $\alpha$	    | греческая буква|
|\infty	      | $\infty$	    |  бесконечность |

#### Основные правила LaTeX для формул:
* ^ - верхний индекс (степень)

* _ - нижний индекс

* { } - группировка (например, x^{10})

* $$ - для отдельной строки (display mode)

* $ - для встроенной формулы (inline mode)

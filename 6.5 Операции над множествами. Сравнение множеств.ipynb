## 6.5 Операции над множествами. Сравнение множеств

#### Операции над множествами:
* пересечение множеств '&'
* объединение множеств  '|'
* вычитание множеств  '-'
* вычисление симметричной разности  '^'

```python
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6, 7}
print(setA)
print(setB)
```

#### Пересечение множеств. Возвращает общие значения для двух множеств.
#### При этом исходные множеств остаются без изменений.

```python
setA & setB
```

```python
res = setA & setB
print(res)
```

Также можно записать так (присваивание результата пересечения в setA):

```python
setA &= setB
print(setA)
```

Если взять множество которое не имеет общих значений с другим множеством, \
то результатом будет пустое множество.

```python
setC = {9, 10, 11}
print(setA & setC)
```

Вместо оператора пересечения '&' можно использовать специальный метод \
intersection()

```python
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6, 7}
setC = setA.intersection(setB)
print(setC)
```

```python
print(setA)
setA.intersection_update(setB)
print(setA)
```

#### Объединение множеств '|'

```python
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6, 7}
print(setA | setB)
```

На выходе получаем новое множество, состоящее из всех значений \
обоих множеств, без их дублирования.

#### Метод для объединения union()

```python
setA.union(setB)
```

```python
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6, 7}
setA |= setB
print(setA)
```

## Вычитание множеств '-'

```python
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6, 7}
print(setA - setB)
```

Возвратит новое множество, в котором из множества setA будут удалены все значения \
повторяющиеся в множестве setB:

```python
print(setB - setA)
```

```python
setA -= setB
setB -= setA
print(setA)
print(setB)
```

#### Симметричная разность '^'
#### Возвращает множество, состоящее из неповторяющихся значений обоих множеств. 

```python
setA = {1, 2, 3, 4}
setB = {3, 4, 5, 6, 7}
print(setA ^ setB)
```

#### Сравнение множеств

Множества считаются равными, если все элементы, входящие в одно из них, \
также принадлежат другому и они содержат одинаковое число элементов.

```python
setA = {7, 6, 5, 4, 3}
setB = {3, 4, 5, 6, 7}
print(setA == setB)
```

```python
print(setA != setB)
```

```python
setA = {7, 6, 5, 4, 3}
setB = {3, 4, 5}
print(setB < setA)
```

```python
print(setB > setA)
```

Если хотя бы один элемент множества setB не будет принадлежать множеству setA \
то обе операции вернут False для равных множеств.

```python
setB.add(22)
print(setA)
print(setB)
```

```python
print(setB > setA)
```

```python
print(setB < setA)
```

Операторы <= или >= вернут True

```python
setA = {7, 6, 5, 4, 3}
setB = {3, 4, 5, 6, 7}
print(setA <= setB)
```

```python
setB.add(1)
print(setB)
print(setA <= setB)
```

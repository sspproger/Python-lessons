## 7.6 Операторы упаковки и распаковки коллекций

```python
x, y = (1, 2)
print(x, y)
```

```python
x, y = 1, 2
print(x, y)
```

```python
x, y = (1, 2, 3, 4)
```

```python
x, *y = (1, 2, 3, 4)
print(x)
print(y)
```

```python
*x, y = (1,2, 3, 4)
print(f"x = {x}")
print(f"y = {y}")
```

```python
x, *y = [1, "a", True, 4]
print(f"x = {x}")
print(f"y = {y}")
```

```python
*x, y, z = "Hello Python!"
print(f"x = {x}")
print(f"y = {y}")
print(f"z = {z}")
```

Упакованные элементы упаковать нельзя!

```python
*y = 1, 2, 3
```

```python
x, *y = 1, 2, 3
print(f"x = {x}")
print(f"y = {y}")
```

Оператором ' * ' можно выполнять и обратную операцию - \
распаковывать коллекции в наборы данных.

```python
a = [1, 2, 3]
print(f"Упаковано = {(a,)}")
print(f"Распаковано = {(*a,)}") 
```

```python
d = -5, 5
print(range(d))
```

```python
print(range(*d))
```

```python
print(list(range(*d)))
```

```python
[*range(*d)]
```

Таким образом можно делать объединение разных коллекций в одну коллекцию, \
например список:

```python
[*range(*d), *(True, False), *a]
```

```python
d = {0: "безнадёжно", 1: "убого", 2: "неуд.", 3: "удовл.", 4: "хорошо", 5: "отлично"}
print({*d}) #множество из ключей
```

```python
print([*d]) # список из ключей
```

```python
print([*d.values()]) # список из значений словаря d
```

```python
print([*d.items()]) # список кортежей из словаря d пары ключ-значение
```

Если требуется распокавать словарь как словарь, \
то перед ним следует прописать две звездочки:

```python
print({**d}) # словарь
```

```python
d2 = {6: "превосходно", 7: "элитарно", 8: "божественно"}
```

Соедим два словаря d и d2 через распаковку:

```python
print({**d, **d2})
```

Оператор ' ** ' две звёздочки используется только для распаковки СЛОВАРЯ!
Для упаковки оператор две звёздочки используется только в параметрах функции, \
так **kwargs чтобы упаковать именнованные параметры функции.

```python
x, *t, z = (1, 2, 3, 4, 5)
print(t)
```

```python
a = [1, 2, 3]
print(*a)
```

```python
*lst, x, y, z = map(int, input().split())
print(*lst)
```

```python
lst_c = (*input().split(),)
print(lst_c)
```

```python
a, b = map(int, input().split())
lst = [*range(a,b+1)]
print(lst)
print(*lst)
```

```python
numbers = map(float, input().split())
cities = input().split()
lst = [*numbers, *cities]
print(*lst)
```

```python
menu = {'Главная': 'home', 'Архив': 'archive', 'Новости': 'news'}
# lst_in = list(map(str.strip, sys.stdin.readlines()))
#d = dict(Города="about-cities", Машины="read-of-cars", Самолеты="airplanes")
d = ["Города=about-cities", "Машины=read-of-cars", "Самолеты=airplanes"]
menu = {**menu, **dict(i.split('=') for i in d)}
print(menu)
```

## 4.3 Тернарный условный оператор

#### <значение 1> if <условие> else <значение 2>

```python
a = 12
b = 7
if a > b:
    res = a
else:
    res = b
print(res)
```

```python
res = a if a > b else b
print(res)
```

### Тернарный оператор автоматически возвращает результат. В данном примере – это или переменная a или переменная b. Тогда как обычный условный оператор предназначен для выполнения блока кода по определенному условию и сам по себе не возвращает никаких значений. 

```python
a = 3 
b = 8
res = a + 2 if a < b else b - 5
print(res)
```

```python
a = -12
b = -7
res = abs(a) if a > b else abs(b)
print(res)
```

```python
res = print(a) if a > b else print(b)
```

```python
s = 'python'
t = 'upper'
res = s.upper() if t == 'upper' else s
print(res)
```

```python
[1, 2, a if a < b else b, 4, 5]
```

```python
a = 8
print("a - " + ("четное" if a % 2 == 0 else "нечетное") + " число")
```

```python
a = 7
print("a - " + ("четное" if a % 2 == 0 else "нечетное") + " число")
```

## Вложенный тернарный оператор

```python
a = 2
b = 3
c = -4
d = (a if a > c else c) if a > b else (b if b > c else c)
print(d)
```

```python
a = float(input())
b = float(input())
d = a if a > b else b
print(d)
```

```python
n = int(input())
msg = "кратно 3" if n % 3 == 0 else "не кратно 3"
print(msg)
```

```python
s1 = input().lower()
s2 = s1[::-1]
print( "палиндром" if s1 == s2 else "не палиндром")
```

```python
n = int(input())
print("False" if n == 0 else "True")
```

```python
n = int(input())
res = n + 1 if n < 59 else 0
print(res)
```

```python
m = ['#до', 'ре', 'ми', '#фа', 'соль', 'ля', 'си']
a, b, c = map(int, input().split())
print(f"{m[a-1]} {m[b-1]} {m[c-1]}")
```

```python
m = ['до', 'ре', 'ми', 'фа', 'соль', 'ля', 'си']
a, b, c = map(int, input().split())
print(('#' if a in (1,4) else '') + m[a - 1] + ' ' + ('#' if b in (1,4) else '') + m[b-1] + ' ' + ('#' if c in (1,4) else '') + m[c-1])
```
